```python
def load_custom_conll(file_path):
    data = []
    with open(file_path, "r", encoding="utf-8") as f:
        current_tokens = []
        current_labels = []
        for line in f:
            line = line.strip()
            if line:
                parts = line.split()  # Split by whitespace
                if len(parts) >= 2:  # Ensure there's at least a token and label
                    token = parts[0]
                    label = parts[-1]  # Assume last part is the label
                    current_tokens.append(token)
                    current_labels.append(label)
            else:
                if current_tokens:
                    data.append((current_tokens, current_labels))
                    current_tokens = []
                    current_labels = []
        # Add the last sentence if file doesn't end with a blank line
        if current_tokens:
            data.append((current_tokens, current_labels))
    return data

# Usage
data = load_custom_conll("../data/processed/labeled_data.txt")
print(data[:2])  # Check first 2 entries
```
